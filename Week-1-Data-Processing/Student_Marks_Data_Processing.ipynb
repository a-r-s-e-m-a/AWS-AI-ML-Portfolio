```python
import csv
import numpy as np
```

```python
data = [
    ["name", "mark"],
    ["  Hana  ", "85"],
    ["Abel", "72"],
    [" Sara ", "91"],
    ["Miki", ""],
    [" Dawit ", "45"],
    ["Liya", "110"],
    ["Meron", "60"]
]

with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(data)

print("CSV file created successfully!")
```

```python
def load_data(filename):
    with open(filename, "r", newline="") as file:
        reader = csv.DictReader(file)
        data = list(reader)

    return data


raw_data = load_data("students.csv")

print("Original data:")
print(raw_data)
```

```python
def clean_name(name):
    return name.strip()


def clean_mark(mark):
    try:
        mark = float(mark)

        if 0 <= mark <= 100:
            return mark

        return None

    except ValueError:
        return None
```

```python
def clean_data(data):
    cleaned_data = []
    removed_records = 0

    for row in data:
        name = clean_name(row["name"])
        mark = clean_mark(row["mark"])

        if name != "" and mark is not None:
            cleaned_data.append({
                "name": name,
                "mark": mark
            })
        else:
            removed_records += 1

    return cleaned_data, removed_records


cleaned_data, removed_records = clean_data(raw_data)

print("Cleaned data:")
for student in cleaned_data:
    print(student)

print("Removed records:", removed_records)
```

```python
def process_data(data):
    marks = np.array([student["mark"] for student in data])

    average = np.mean(marks)
    highest = np.max(marks)
    lowest = np.min(marks)
    passed = np.sum(marks >= 50)

    return average, highest, lowest, passed


average, highest, lowest, passed = process_data(cleaned_data)

print("Number of valid students:", len(cleaned_data))
print("Average mark:", round(average, 2))
print("Highest mark:", highest)
print("Lowest mark:", lowest)
print("Number of students who passed:", passed)
print("Number of students who failed:", len(cleaned_data) - passed)
```

```python
print("=" * 35)
print("STUDENT MARKS DATA REPORT")
print("=" * 35)

print("Total records loaded:", len(raw_data))
print("Valid student records:", len(cleaned_data))
print("Removed records:", removed_records)
print("Average mark:", round(average, 2))
print("Highest mark:", highest)
print("Lowest mark:", lowest)
print("Students passed:", passed)
print("Students failed:", len(cleaned_data) - passed)

print("=" * 35)
print("PROCESSING COMPLETED SUCCESSFULLY!")
```

# Student Marks Data Processor

## 1. Introduction

The Student Marks Data Processor is a small Python application that loads student marks from a CSV file, cleans invalid data, and calculates statistics using NumPy.

## 2. Objectives

- Load a dataset from a CSV file.
- Clean student names and marks using Python functions.
- Handle missing and invalid marks.
- Use NumPy to calculate statistical results.
- Run the application in a Jupyter notebook.

## 3. Tools and Technologies

- Python: Main programming language.
- CSV: File format for storing student records.
- NumPy: Library used for numerical processing.
- Google Colab: Environment used to run the notebook.
- venv: Virtual environment for managing project dependencies.
- requirements.txt: File listing the required Python packages.

## 4. Data Cleaning Rules

- Remove unnecessary spaces from student names.
- Convert valid marks into numbers.
- Accept marks from 0 to 100.
- Remove records with empty names or missing or invalid marks.

## 5. Processing

The program calculates the number of valid students, average mark, highest mark, lowest mark, number of students who passed, and number who failed. A mark of 50 or above is considered a pass.

## 6. Conclusion

The project demonstrates how Python functions, CSV files, and NumPy can be used together to load, clean, and process a small dataset.
